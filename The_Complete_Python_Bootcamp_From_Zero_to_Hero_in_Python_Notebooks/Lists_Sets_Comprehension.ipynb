# zip()
Pair up two or more lists item by item and returns a tuple

```python
mylist1 = ['1','2','3','4','5']
mylist2 = ['chr']*5
```

```python
myzip_tuple = zip(mylist1,mylist2)

# cast zip object to a list
print(list(myzip_tuple))
```

# Dictionaries

```python
mydict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(mydict.keys())
print(mydict.values())
```

```python
# iterate over a dictionary keys
for key in mydict:
    print(key)

# iterate over a dictionary values
for value in mydict.values():
    print(value)
```

```python
# unpack a dictionary

for key, value in mydict.items():
    print(f"{key}: {value}")
```

```python
# check if dictionary contains a key
print('brand' in mydict) 

# check if dictionary contains a value
print(1964 in mydict.values()) 
```

# enumerate()

```python
index_count = 0
for letter in 'abcde':
    print(f"Index {index_count}: {letter}")
    index_count += 1
```

The enumerate function adds a counter to an iterable (like a list, tuple, or string) and returns it as an enumerate object, which yields tuples containing the count and the item. This eliminates the need to manually initialize and increment loop counters, making your code cleaner and more Pythonic

```python
word = 'abcde'

for index, letter in enumerate(word):
    print(f"Index {index}: {letter}")
```

# List Comprehension

```python
mystring = 'hello'

# for loop
mylist = []
for letter in mystring:
    mylist.append(letter)
mylist
```

List comprehension is a flat version of a for loop 
* [expression for item in iterable if condition]
* default expression is append

```python
mylist = []
mylist = [letter for letter in mystring]
mylist
```

```python
my_number_list = []
my_number_list = [number for number in range(0,11)]
my_number_list
```

```python
my_number_list = []
my_number_list = [number**2 for number in range(0,11)]
my_number_list
```

```python
even_number_list = []
even_number_list = [number**2 for number in range(0,11) if number%2 == 0]
even_number_list
```

```python
celcius = [0,10,20,34.5]
fahrenheit = []
fahrenheit = [((9/5) * temp + 32) for temp in celcius]
fahrenheit
```

```python
# if/else

result = [number if number%2 == 0 else 'ODD' for number in range(0,11)]
result
```

```python
# nested for loop
mylist = []
for x in [2,4,5]:
    for y in [1,10,100]:
        mylist.append(x*y)
```

```python
# nested list comprehension loop
mylist = [x*y for x in [2,4,6] for y in [1,10,100]]
mylist
```
