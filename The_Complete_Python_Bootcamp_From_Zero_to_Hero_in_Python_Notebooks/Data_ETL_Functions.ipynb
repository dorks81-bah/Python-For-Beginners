# 5. map()
### Useful for applying a transformation to each item of a sequence.
* Combine with custom function
* Combine with lambda function
* Combine with if-elif-else
* Multiple lists

```python
# Converting strings to integers

str_numbers = ["1", "2", "3", "4"]
int_numbers = list(map(int, str_numbers))
print(int_numbers)  # Output: [1, 2, 3, 4]
```

```python
# Combine with custom function

def double(x):
  return x * 2


numbers = [10, 20, 30]
doubled_numbers = list(map(double, numbers))
print(doubled_numbers)  # Output: [20, 40, 60]
```

```python
def categorize_temp(temp):
    if temp >= 80:
        return "Hot"
    elif temp >= 60:
        return "Mild"
    else:
        return "Cold"

temps = [55, 72, 88, 61]
labels = list(map(categorize_temp, temps))

print(labels)  # Output: ['Cold', 'Mild', 'Hot', 'Mild']
```

```python
# Combine with lambda function

numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))
print(squared)  # Output: [1, 4, 9, 16]
```

```python
# if-else

numbers = [1, 2, 3, 4, 5, 6]

# Double even numbers, leave odd numbers unchanged
result = list(map(lambda x: x * 2 if x % 2 == 0 else x, numbers))

print(result)
# Output: [1, 4, 3, 8, 5, 12]
```

```python
# if-elif-else alternative to categorize_temp(temp) above

temps = [55, 72, 88, 61]
labels2 = list(map(lambda temp: "Hot" if temp >= 80 else ("Mild" if temp >= 60 else "Cold"), temps))

print(labels2)  # Output: ['A', 'B', 'F', 'B']
```

```python
# Multiple lists

list_a = [1, 2, 3]
list_b = [10, 20, 30]

sums = list(map(lambda x, y: x + y, list_a, list_b))
print(sums)  # Output: [11, 22, 33]
```
