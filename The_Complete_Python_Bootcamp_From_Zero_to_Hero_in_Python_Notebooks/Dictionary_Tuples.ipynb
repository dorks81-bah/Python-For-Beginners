# Dictionary
1. dict.keys() and dict.values()
2. dict.items()
3. dict.get()
4. dict.pop(), dict.update(), del

```python
mydict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
mydict
```

## 1. dict.key() and dict.values() returns a dict_key or dict_values data type. This class can then be cast to a list of keys or values

```python
mykeys = mydict.keys()
myvalues = mydict.values()
print(mykeys)
print(myvalues)
print_var_type(mykeys = mykeys,
               myvalues = myvalues)

# cast to a list

keys_list = list(mydict.keys())
values_list = list(mydict.values())
print(keys_list)
print(values_list)
print_var_type(keys_list = keys_list,
               values_list = values_list)
```

```python
# iterate over a dictionary keys

for key in mydict:
    print(key)

# iterate over a dictionary values

for value in mydict.values():
    print(value)
```

## 2. dict.items() returns a dict_items data type which is list of key-value tuples

```python
print(mydict.items())
print(type(mydict.items()))
print(type(list(mydict.items())))
```

```python
# iterate using dict.items()

for x, y in mydict.items():
  print(x, y)
```

## 3. dict.get() is a cleaner and safer method for retrieving values 

```python
print(mydict['brand'])  # bad error handling
print(mydict.get('brand'))  # cleaner error handling
```

## 4. Changing a dictionary

#### dict.update()

```python
# both methods work but dict.update() is safer

mydict['year'] = 2020
print(mydict.get('year')) 
mydict.update({"year": 2026})
print(mydict.get('year')) 
```

#### dict.pop and del

```python
mydict.pop('year')
print(mydict)

del mydict['model']
print(mydict)
```

#### adding a key-value pair with dict.update()

```python
# both work but dict.update() is safer

mydict["year"] = 2026
mydict.update({"color": "red"})
print(mydict)
```

```python
def myfunc(input:str) -> str:
    transformed = ''
    index = 0
    for index, letter in enumerate(input):
        if index == 0:
            transformed += letter.lower()
        elif index%2 == 0:
            transformed += letter.lower()
        else:
            transformed += letter.upper()
    return transformed

print(myfunc('Anthropomorphism'))
```
