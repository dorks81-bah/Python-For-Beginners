**Write a function that computes the volume of a sphere given its radius.**
<p>The volume of a sphere is given as $$\frac{4}{3} πr^3$$</p>

```python
def vol(rad: int|float) -> float:
    #pass
    return (4/3)*3.14*rad**3
```

**Write a function that checks whether a number is in a given range (inclusive of high and low)**

```python
def ran_check(num: int,low: int,high: int) -> None:
    # pass
    if num >= low and num <= high:
        print(f'{num} is in the range between {low} and {high}')
    else:
        print(f'{num} is not in the range between {low} and {high}')
```

```python
# Check
ran_check(5,2,7)  # 5 is in the range between 2 and 7
ran_check(2,2,7)  # 2 is in the range between 2 and 7
ran_check(7,2,7)  # 7 is in the range between 2 and 7
ran_check(0,2,7)  # 0 is not in the range between 2 and 7
```

If you only wanted to return a boolean:

```python
def ran_bool(num: int,low: int,high: int) -> bool:
    #pass
    if num >= low and num <= high:
        return True
    else:
        return False
```

```python
print(ran_bool(3,1,10))  # True
print(ran_bool(1,1,10))  # True
print(ran_bool(10,1,10))  # True
print(ran_bool(0,1,10))  # False
```

**Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.**

    Sample String : 'Hello Mr. Rogers, how are you this fine Tuesday?'
    Expected Output : 
    No. of Upper case characters : 4
    No. of Lower case Characters : 33

HINT: Two string methods that might prove useful: **.isupper()** and **.islower()**

If you feel ambitious, explore the Collections module to solve this problem!

```python
def up_low(s: str) -> None:
    #pass
    upper = 0
    lower = 0
    for l in s:
        if l.isupper():
            upper += 1
        elif l.islower():
            lower += 1
    print(f'Original String: {s}')
    print(f'No. of Upper case characters :  {upper}')
    print(f'No. of Lower case Characters :  {lower}')
    
```

```python
s = 'Hello Mr. Rogers, how are you this fine Tuesday?'
up_low(s)

# Original String: Hello Mr. Rogers, how are you this fine Tuesday?
# No. of Upper case characters :  4
# No. of Lower case Characters :  33
```

**Write a Python function to multiply all the numbers in a list.**

    Sample List : [1, 2, 3, -4]
    Expected Output : -24

```python
def multiply(numbers: int):  
    #pass
    result = 1
    for num in numbers:
        result *= num

    print(result)
```

```python
multiply([1,2,3,-4])  # -24
```

**Write a Python function that checks whether a word or phrase is palindrome or not.**

Note: A palindrome is word, phrase, or sequence that reads the same backward as forward, e.g., madam,kayak,racecar, or a phrase "nurses run". Hint: You may want to check out the .replace() method in a string to help out with dealing with spaces. Also google search how to reverse a string in Python, there are some clever ways to do it with slicing notation.

```python
def palindrome(s: str) -> bool:
    #pass
    # assume s contains only alpha numerical text a set to lower case 
    cleaned = "".join(char for char in s.casefold())
    
    # Compare the string to its reverse
    return cleaned == cleaned[::-1]
```

```python
print(palindrome('helleh'))  # True
print(palindrome('racecar'))  # True
print(palindrome('Hello World'))  # False
```

**Write a Python function to check whether a string is pangram or not. (Assume the string passed in does not have any punctuation)**

    Note : Pangrams are words or sentences containing every letter of the alphabet at least once.
    For example : "The quick brown fox jumps over the lazy dog"

Hint: You may want to use .replace() method to get rid of spaces.

Hint: Look at the [string module](https://stackoverflow.com/questions/16060899/alphabet-range-in-python)

Hint: In case you want to use [set comparisons](https://medium.com/better-programming/a-visual-guide-to-set-comparisons-in-python-6ab7edb9ec41)

```python
import string

def ispangram(str1: str, alphabet=string.ascii_lowercase) -> bool:
    #pass
    clean = (str1.replace(" ","")).lower()
    return set(clean) == set(alphabet)
```

```python
print(ispangram("The quick brown fox jumps over the lazy dog"))  # True
print(ispangram("The quick red fox jumps over the lazy dog"))  # False# True
```
