# 2. Returning and unpacking multiple values

```python
def display_profile(name: str, age: int, role: str) -> str:
    name = name
    age = age
    role = role
    
    return name, age, role

# returns a tuple
info = display_profile("Alice",30,"Engineer")

# returns tuple unpacked
name, age, role = display_profile("Alice",30,"Engineer")

print(info)
print(name)
print(age)
print(role)
```

# 3. *args
### Allows a function to accept any number of positional arguments as a tuple

```python
def multiply_all(*args):
    # 'args' is treated as a tuple inside the function
    result = 1
    for num in args:
        result *= num
    return result

# You can pass 2, 3, or 0 arguments:
print(multiply_all(2, 3))        # Output: 6
print(multiply_all(2, 3, 4, 5))  # Output: 120
print(multiply_all())
```

# 4. **kwargs
### Allows a function to accept an arbitrary number of keyword (named) arguments as a dictionary

```python
def display_profile(**kwargs):
    print(kwargs)  # It's just a dictionary!
    
    # You can loop through it
    for key, value in kwargs.items():
        print(f"{key}: {value}")

display_profile(name="Alice", age=30, role="Engineer")
```

### *args and  **kwargs used in the same function

```python
def order_food(*args, **kwargs):
    # *args collects positional arguments into a tuple
    print("Items ordered:")
    for item in args:
        print(f"- {item}")
        
    # **kwargs collects keyword arguments into a dictionary
    print("\nSpecial delivery instructions:")
    for key, value in kwargs.items():
        print(f"- {key.replace('_', ' ').title()}: {value}")

# Example usage:
order_food("Burger", "Fries", "Shake", delivery_time="8:00 PM", leave_at_door=True)
```

# 5. lambda functions
### A lambda function is a small anonymous function.
### A lambda function can take any number of arguments, but can only have one expression.
### A lambda function is commonly used inside another function.

```python
def myfunction(a:int) -> int:
    print(a + 10)

# a lambda function resembles myfunction as a single line i.e., a: a + 10
x = lambda a : a + 10
print(x(5))
myfunction(5)
```

```python
# a function can return a lambda function
# return_lambda_function returns the following lambda function
# lambda a : a * n; when n equals to the input of return_lambda_function(n)
def return_lambda_func(n:int) -> int:
    return lambda a : a * n
```

```python
mydoubler = return_lambda_func(2)  # x = lambda a: a * 2
print(mydoubler(11))  # 11 * 2 = 22
```

```python
mytripler = return_lambda_func(3)  # x = lambda a: a * 3
print(mytripler(11))  # 11 * 3 = 33
```
