```python
print(f'{1 < 2 and 3 < 4}')
print(f'{1 < 2 and 4 < 3}')
print(f'{1 < 2 & 3 < 4}')
print(f'{1 < 2 & 4 < 3}')
print(f'{not(1 < 2 & 4 < 3)}')
print(f'{not 1 < 2 & 4 < 3}')
```

```python
# The pipe symbol | is the bitwise OR operator (not a logical OR)
# Comparison operators like < are chained: 1 < 2 | 4 < 3 is parsed as a chained comparison equivalent to:
# (1<(2|4)) and ((2|4)<3)
# Both parts must be true for a bitwise comparrsion
print(f'{1 < 2 or 4 < 3}')
print(f'{1 < 2 | 4 < 3}')
print(f'{not(1 < 2 | 4 < 3)}')
```
